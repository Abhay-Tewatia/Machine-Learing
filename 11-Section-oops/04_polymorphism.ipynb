```python
# Method Overriding — Depth Me
# Parent ne ek method banaya
# Child ne wahi method apne hisaab se dobara likha
# → Yahi Method Overriding hai

class Animal:
    def speak(self):
        print("Some sound...")   # Parent ka version

class Dog(Animal):
    def speak(self):
        print("Woof!")           # Child ne override kiya ✅

class Cat(Animal):
    def speak(self):
        print("Meow!")           # Child ne override kiya ✅

a = Animal()
d = Dog()
c = Cat()

a.speak()    # Some sound...  ← Parent ka chala
d.speak()    # Woof!          ← Dog ka chala
c.speak()    # Meow!          ← Cat ka chala
```

```python
# Case 1 — Sirf Override karo
class Shape:
    def area(self):
        print("Area calculate karo")   # generic

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):                    # Override ✅
        result = 3.14 * self.radius * self.radius
        print(f"Circle area: {result}")

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length  = length
        self.breadth = breadth

    def area(self):                    # Override ✅
        result = self.length * self.breadth
        print(f"Rectangle area: {result}")


# Objects
s = Shape()
c = Circle(5)
r = Rectangle(4, 6)

s.area()    # Area calculate karo    ← Parent
c.area()    # Circle area: 78.5     ← Circle ka formula
r.area()    # Rectangle area: 24    ← Rectangle ka formula
```

```python
# Case 2 — Override + super() bhi use karo

# Kabhi kabhi child ko:
# Parent ka kaam bhi karna hota hai +
# Apna kuch extra bhi karna hota hai
# → Tab super() use karte hain
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name   = name
        self.salary = salary

    def show_details(self):
        # Parent ka basic info
        print(f"Name:   {self.name}")
        print(f"Salary: Rs.{self.salary}")


class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)
        self.team_size = team_size

    def show_details(self):          # Override ✅
        super().show_details()       # Parent ka kaam bhi karo
        # Apna extra kaam
        print(f"Team:   {self.team_size} members")
        print(f"Bonus:  Rs.{self.salary * 0.2}")


class Intern(Employee):
    def __init__(self, name, salary, duration):
        super().__init__(name, salary)
        self.duration = duration

    def show_details(self):          # Override ✅
        super().show_details()       # Parent ka kaam bhi karo
        # Apna extra kaam
        print(f"Duration: {self.duration} months")


# Objects
e = Employee("Ramesh", 50000)
m = Manager ("Abhay",  80000, 5)
i = Intern  ("Priya",  10000, 3)

print("── Employee ──")
e.show_details()
# Name:   Ramesh
# Salary: Rs.50000

print(" \n---Manager---")
m.show_details()
# Name:   Abhay          ← super() se aaya
# Salary: Rs.80000       ← super() se aaya
# Team:   5 members      ← Manager ka apna
# Bonus:  Rs.16000.0     ← Manager ka apna

print("\n── Intern ──")
i.show_details()
# Name:     Priya        ← super() se aaya
# Salary:   Rs.10000     ← super() se aaya
# Duration: 3 months     ← Intern ka apna
```

```python
# Case 3 — Override without super()
# Kabhi kabhi child ko parent ka
# kuch bhi nahi chahiye — poora naya kaam karna hai
# → Tab super() nahi lagate
```

```python
class Notification:
    def send(self, message):
        print(f"Sending: {message}")    # generic


class EmailNotification(Notification):
    def __init__(self, email):
        self.email = email

    def send(self, message):            # Poora override — super() nahi ✅
        print(f"📧 Email to {self.email}")
        print(f"   Message: {message}")
        print(f"   Status: Delivered ✅")


class SMSNotification(Notification):
    def __init__(self, phone):
        self.phone = phone

    def send(self, message):            # Poora override — super() nahi ✅
        print(f"📱 SMS to {self.phone}")
        print(f"   Message: {message}")
        print(f"   Status: Sent ✅")


class WhatsAppNotification(Notification):
    def __init__(self, phone):
        self.phone = phone

    def send(self, message):            # Poora override — super() nahi ✅
        print(f"💬 WhatsApp to {self.phone}")
        print(f"   Message: {message}")
        print(f"   Status: Delivered ✅")


# Ek hi function — sab handle karta hai
def notify_user(notification, message):
    notification.send(message)          # Polymorphism ✅


email    = EmailNotification   ("abhay@gmail.com")
sms      = SMSNotification     ("+91-9996750784")
whatsapp = WhatsAppNotification("+91-9996750784")

notify_user(email,    "Your order is placed!")
# 📧 Email to abhay@gmail.com
#    Message: Your order is placed!
#    Status: Delivered ✅

notify_user(sms,      "OTP: 4829")
# 📱 SMS to +91-9996750784
#    Message: OTP: 4829
#    Status: Sent ✅

notify_user(whatsapp, "Interview at 3PM tomorrow")
# 💬 WhatsApp to +91-9996750784
#    Message: Interview at 3PM tomorrow
#    Status: Delivered ✅
```
