```python
# Inheritance in Python

# Parent Class (Base)  →  jo deta hai
# Child Class (Derived) →  jo leta hai + apna kuch add karta hai
```

```python
# Inheritance
# Parent class
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"The person will drive the {self.enginetype} car")

car1=Car(4,5,"petrol")    
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, is_selfdriving):
        super().__init__(windows, doors, enginetype)
        self.is_selfdriving = is_selfdriving
    def selfdriving(self):
         print(f"Tesla supports self driving: {self.is_selfdriving}")

tesla1= Tesla(4,5, "electric", True)
tesla1.drive()
tesla1.selfdriving()
```

```python
tesla1.drive()
```

```python
class Animal:        # Parent
    def breathe(self):
        print("Breathing...")

class Dog(Animal):   # Child — Animal ki sab cheezein mil gayi
    def bark(self):
        print("Woof!")

dog = Dog()
dog.breathe()        # Parent ka method — bina likhe kaam kiya ✅
dog.bark()           # Apna method
```

```python
# one more example of  single inheritance
class Doctor:
    def __init__(self, name, experience, fees):
        self.name       = name
        self.experience = experience
        self.fees       = fees
        self.patients   = []   # treated patients list
    
    def consult(self, patient_name):
        self.patients.append(patient_name)
        print(f"Dr. {self.name} is consulting {patient_name}")

    def show_patients(self):
        if not self.patients:
            print("No patient yet")
        else:
            print(f"Dr. {self.name}'s Patients: {self.patients}")
    
    def __str__(self):
        return f"Dr. {self.name} | {self.experience} yrs exp | Rs.{self.fees}/visit"

class Surgeon(Doctor):
    def __init__(self, name, experience, fees, specialization):
        super().__init__(name, experience, fees)
        self.specialization = specialization
        self.surgeries_done = 0           # apna extra data

    def perform_surgery(self, patient_name):
        self.surgeries_done += 1
        print(f"Dr. {self.name} performing {self.specialization} surgery on {patient_name} 🏥")
        print(f"Total surgeries done: {self.surgeries_done}")

    def __str__(self):
        return f"{super().__str__()} | {self.specialization} Surgeon"

class Psychiatrist(Doctor):
    def __init__(self, name, experience, fees, therapy_type):
        super().__init__(name, experience, fees)
        self.therapy_type    = therapy_type
        self.sessions_done   = 0

    def conduct_therapy(self, patient_name):
        self.sessions_done += 1
        print(f"Dr. {self.name} conducting {self.therapy_type} session with {patient_name} 🧠")
        print(f"Total sessions done: {self.sessions_done}")

    def __str__(self):
        return f"{super().__str__()} | {self.therapy_type} Therapist"


surgeon = Surgeon("Rohan Mehta", 15, 2000, "Cardiac")
psych   = Psychiatrist("Priya Sharma", 8, 1500, "Cognitive Behavioral")

print(surgeon)
print(psych)

```

```python
# parent ka method dono use kr rhe hai
surgeon.consult("Abhay")
surgeon.consult("Rahul")
psych.consult("rogu")
surgeon.show_patients()         
psych.show_patients()         
```

```python
print()

# Apna exclusive method
surgeon.perform_surgery("Abhay")  
surgeon.perform_surgery("Rahul") 
print()

psych.consult("Priya")        
psych.conduct_therapy("Priya")  
psych.conduct_therapy("Neha")  
psych.show_patients()  
```

```python
# multiple inheitance
# when a class inherits from more than one base class
# Parent A     Parent B
#     \           /
#      \         /
#       \       /
#       Child (dono se leta hai)
```

```python
# ── PARENT 1 ─────────────────────────────────────────
class Batsman:
    def __init__(self, batting_avg, centuries):
        self.batting_avg = batting_avg
        self.centuries   = centuries
        self.runs        = 0

    def bat(self):
        print(f"Taking stance at crease 🏏")

    def score_runs(self, runs):
        self.runs += runs
        print(f"Scored {runs} runs! Total: {self.runs}")

    def show_batting(self):
        print(f"Batting Avg: {self.batting_avg} | Centuries: {self.centuries} | Runs: {self.runs}")


# ── PARENT 2 ─────────────────────────────────────────
class Bowler:
    def __init__(self, bowling_speed, wickets):
        self.bowling_speed = bowling_speed
        self.wickets       = wickets

    def bowl(self):
        print(f"Running in to bowl at {self.bowling_speed} km/h 🎳")

    def take_wicket(self, batsman_name):
        self.wickets += 1
        print(f"WICKET! {batsman_name} is OUT! Total wickets: {self.wickets}")

    def show_bowling(self):
        print(f"Bowling Speed: {self.bowling_speed} km/h | Wickets: {self.wickets}")


# ── CHILD — Dono se le raha hai ──────────────────────
class AllRounder(Batsman, Bowler):
    def __init__(self, name, batting_avg, centuries, bowling_speed, wickets, country):
        # Dono parents ka __init__ call karna padega
        Batsman.__init__(self, batting_avg, centuries)
        Bowler.__init__ (self, bowling_speed, wickets)
        # Apna
        self.name    = name
        self.country = country

    def perform(self):
        print(f"\n{self.name} ({self.country}) — All Round Performance:")
        print(f"  ", end=""); self.show_batting()
        print(f"  ", end=""); self.show_bowling()

    def match_winning_performance(self, runs, batsman_out):
        print(f"\n⭐ {self.name} — Match Winning Performance!")
        self.score_runs(runs)            # Batsman ka method
        self.take_wicket(batsman_out)    # Bowler ka method

    def __str__(self):
        return f"{self.name} | {self.country} | AllRounder"


# ── OBJECTS ──────────────────────────────────────────
hardik = AllRounder(
    name          = "Hardik Pandya",
    batting_avg   = 35.4,
    centuries     = 0,
    bowling_speed = 140,
    wickets       = 85,
    country       = "India"
)

jadeja = AllRounder(
    name          = "Ravindra Jadeja",
    batting_avg   = 37.2,
    centuries     = 3,
    bowling_speed = 118,
    wickets       = 280,
    country       = "India"
)

print(hardik)        
print(jadeja)        

# Batsman ka method
hardik.bat()                      
hardik.score_runs(67)          
hardik.score_runs(33)          

# Bowler ka method
hardik.bowl()                
hardik.take_wicket("Steve Smith")   
hardik.take_wicket("David Warner") 

# Dono ka combined
hardik.perform()

hardik.match_winning_performance(55, "Ben Stokes")

jadeja.perform()
```
