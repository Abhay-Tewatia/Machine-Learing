```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} deposited. Balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
        else:
            self.balance -= amount
            print(f"{amount} withdrawn. Balance: {self.balance}")

    def show_balance(self):
        print(f"{self.owner}'s Balance: {self.balance}")


# Objects
acc1 = BankAccount("Abhay", 1000)
acc2 = BankAccount("Rahul")        # balance default 0

acc1.deposit(500)                   # 500 deposited. Balance: 1500
acc1.withdraw(200)                  # 200 withdrawn. Balance: 1300
acc2.deposit(100)                   # 100 deposited. Balance: 100
acc1.show_balance()                 # Abhay's Balance: 1300
acc2.show_balance()
```

```python
class Student:
    # Class variable — sab students ke liye same
    school = "GJUST"

    def __init__(self, name, marks):
        # Instance variable — har student ka apna
        self.name = name
        self.marks = marks

    def grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        else:
            return "C"

    def show(self):
        print(f"{self.name} | Marks: {self.marks} | Grade: {self.grade()} | School: {self.school}")


# Objects
s1 = Student("Abhay", 92)
s2 = Student("Rahul", 78)
s3 = Student("Priya", 65)

s1.show()    # Abhay | Marks: 92 | Grade: A | School: GJUST
s2.show()    # Rahul | Marks: 78 | Grade: B | School: GJUST
s3.show()    # Priya | Marks: 65 | Grade: C | School: GJUST
```

```python
class Phone:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price
        self.is_on = False          # default band hai

    def power_on(self):
        if self.is_on:
            print(f"{self.model} already ON")
        else:
            self.is_on = True
            print(f"{self.model} powered ON ✅")

    def power_off(self):
        self.is_on = False
        print(f"{self.model} powered OFF")

    def info(self):
        status = "ON" if self.is_on else "OFF"
        print(f"{self.brand} {self.model} | Rs.{self.price} | {status}")


# Objects
iphone  = Phone("Apple",   "iPhone 15", 79999)
samsung = Phone("Samsung", "S24",       72999)
oneplus = Phone("OnePlus", "12R",       42999)

iphone.power_on()     # iPhone 15 powered ON
iphone.power_on()     # iPhone 15 already ON
samsung.info()        # Samsung S24 | Rs.72999 | OFF
oneplus.power_on()    # OnePlus 12R powered ON
oneplus.power_off()   # OnePlus 12R powered OFF
```

```python
class JobApplication:
    total_applications = 0          # kitne total applications aaye

    def __init__(self, candidate, role, company):
        self.candidate = candidate
        self.role = role
        self.company = company
        self.status = "Applied"     # default status
        JobApplication.total_applications += 1

    def update_status(self, new_status):
        old = self.status
        self.status = new_status
        print(f"{self.candidate}: {old} → {new_status}")

    def __str__(self):
        return f"{self.candidate} | {self.role} at {self.company} | {self.status}"


# Objects
app1 = JobApplication("Abhay", "Full Stack Dev", "JobLuxe")
app2 = JobApplication("Rahul", "Backend Dev",    "Google")
app3 = JobApplication("Priya", "ML Engineer",    "Amazon")

print(app1)         
print(app2)         

app1.update_status("Interview Scheduled")  
app2.update_status("Rejected")          
app1.update_status("Hired ✅")       

print(f"\nTotal Applications: {JobApplication.total_applications}")  
```
