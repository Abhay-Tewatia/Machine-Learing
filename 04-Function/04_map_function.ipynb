```python
# the map function applies a given function to each item of an iterable (like a list) and returns a map object (which is an iterator). The syntax for the map function is:
# map(function, iterable, ...)
```

```python
def square(x):
    return x*x
```

```python
def square(x):
    return x*x
numbers = [1, 2, 3, 4, 5]
print(list(map(square, numbers)))



```

```python
# lambda function with map function
numbers = [1, 2, 3, 4, 5]
list(map(lambda x:x*x, numbers))
```

```python
# can be map multiple iterables together.
# The function must take as many arguments as there are iterables. The map function will apply the function to the items of the iterables in parallel, and return a map object.
```

```python
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]
numbers3  = [7, 8, 9]

result = list(map(lambda x, y, z : x+y+z, numbers1, numbers2, numbers3))
print(result)
```

```python
# map() to convert a list of string to integrers
str_numbers = ['1', '2', '3', '4', '5']
str_integers = list(map(int, str_numbers))
print(str_integers)
```

```python
words = ['apple', 'banana', 'cherry']
upper_words = list(map(str.upper, words))
print(upper_words)
```

```python
def get_name(person):
    return person['name'], person['age']

people = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Charlie', 'age': 35}
]

list(map(get_name, people))
```
