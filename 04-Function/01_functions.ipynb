```python
def check_even_odd(num):
    if(num%2 ==0):
        print("even")
    else:
        print("odd")

# call the function with different numbers
check_even_odd(24)
check_even_odd(25)
```

```python
# define the function with multiple parameters
def add_numbers(num1, num2):
    return num1 + num2
# call the function and print the result
result = add_numbers(5,3)
print(result)
```

```python
# Default  parameters
def greet(name, greeting = "Hello"):
    return f"{greeting}, {name} welcome to the world of functions!"

call = greet("Abhay")  # Uses default greeting
print(call)
```

```python
# variable length arguments
# positional and keyword arguments

def print_numbers(*args):
    for num in args:
        print(num)

print_numbers(1, 2, 3, 4, 5)        
```

```python
# keyword arguments
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="Abhay", age=25, city="New York")        
```

```python
def print_details(*args, **kwargs):
     for num in args:
        print(f"positional argument: {num}")

     for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(1, 2, 3, name="Abhay", age=25, city="New York")        
```

```python
# return statement multiple values
def multiply_numbers(num1, num2):
    return num1 * num2, num2

result = multiply_numbers(5, 3)
print(result)
```
