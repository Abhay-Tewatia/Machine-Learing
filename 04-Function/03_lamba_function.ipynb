```python
# lambda function is a small anonymous function that can take any number of arguments, but can only have one expression. The syntax for a lambda function is:
# lambda arguments: expression
```

```python
def add(x,y):
    return x+y
add(2,3)
```

```python
addition = lambda a,b: a+b
type(addition)
print(addition(2,3))
```

```python
def even(num):
    return num%2==0

even_lambda = lambda num: num%2==0
print(even_lambda(4))
```

```python
def addi(x,y,z):
    return x+y+z

addi_lambda = lambda x,y,z: x+y+z
print(addi_lambda(1,2,3))
```

```python
list(map(lambda x:x**2, [1,2,3,4,5]))
```
