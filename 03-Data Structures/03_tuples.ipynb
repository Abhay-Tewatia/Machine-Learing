```python
numbers = tuple([1,2,3,4,5,6])
numbers
```

```python
mixed_tuple = (1, "hello", 3.14, [1,2,3])
print(mixed_tuple)
```

```python
# Accessing elements in a tuple
print(mixed_tuple[0])  # Accessing the first element
print(mixed_tuple[1])  # Accessing the second element
```

```python
numbers[0:3]
```

```python
# Tuples are immutable, which means that once a tuple is created, its elements cannot be modified. This immutability makes tuples more memory efficient and faster than lists for certain operations. However, it also means that you cannot add, remove, or change elements in a tuple after it has been created.
```

```python
# Tuple operations

# concatenation
concatenation_tuple = numbers + mixed_tuple
print(concatenation_tuple)
```

```python
mixed_tuple*3
```

```python
numbers*4
```

```python

# immuatable nature of tuples
lst = [1,2,3,4,5]
print(lst)
lst[1] = "Abhay"
print(lst)
```

```python
# tuple methods
numbers
```

```python
print(numbers.count(3))  # Count the number of occurrences of 3
print(numbers.index(3))  # Find the index of the first occurrence of 3
```

```python
# packing and unpacking of tuples
packed_tuple = 1, "hello", 3.14
print(packed_tuple)
```

```python
# unpacking tuple
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
# unpacking with * operator
numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

```python
# nested tuples
lst = [[1,2], [3,4], [1,2,"hello"], [3.14, "c"]]
lst[2][2]
```

```python
nested_tuple = ((1,2), (3,4), (1,2,"hello"), (3.14, "c"))
print(nested_tuple)
print(nested_tuple[2][2])
```

```python
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end = " ")

    print()
```
