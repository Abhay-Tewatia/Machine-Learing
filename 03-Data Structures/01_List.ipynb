<!-- list comprehension -->
basic syntax                [expression for item in iterable]

with conditional logic      [expression for item in iterable if condition]
nested list comprehension   []

```python
# basic syntax  [expression for item in iterable]
square = [nums**2 for nums in range(1, 10)]
print(square)
```

```python
# list comprehension with condition
ssofeven = [x**2 for x in range(1, 20) if x %2 == 0]
print(ssofeven)

```

```python
# one more way to create list usinf loop
new = []
for x in range(1, 10):
    if (x%2 == 0):
        new.append(x**2)
print(new)
```

```python
# nested list comprehension
list1 = [1,2,3,4,5]
list2 = ['a', 'b', 'c', 'd', 'e']
pair = [(i,j) for i in list1 for j in list2]
print(pair)
```

```python
# list comprehension with funciton called len() to get the length of each string in the list
strings = ['hello', 'world', 'python', "Abhay Tewatia"]
lengths = [len(s) for s in strings]
print(lengths)
```

Conclusion

List comprehensions are a powerful and concise way to create lists in Python. They are syntactically compact and can replace more verbose looping constructs. Understanding the syntax of list comprehensions will help you write cleaner and more efficient Python code.
