```python
# sets: sets are used to store multiple items in a single variable. They are unordered, unchangeable, and do not allow duplicate values. Sets are defined using curly braces {} or the set() function.
```

```python
# create a set
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(my_empty_set)
print(type(my_empty_set))
```

```python
my_empty_set = set([1,2,3,4,5])
print(my_empty_set)
```

```python
my_empty_set = set([1,2,3,3,3,4,4,4,4,5])
print(my_empty_set)
```

```python
# basic set operations
A = {1,2,3,4,5}
A.add(6)
print(A)
```

```python
A.remove(4)
print(A)
```

```python
A.discard(11)
print(A)
```

```python
B = {1,2,3,4,5}
B.clear()
print(B)
```

```python
# set membership test
B = {1,2,3,4,5}
print(3 in B)
print(6 in B)
```

```python
# Mathematical operations on set
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
# Union
union_set = set1.union(set2)
print("union_set:", union_set)
```

```python
intersection_set = set1.intersection(set2)
print("intersection_set:", intersection_set)
difference_set = set1.difference(set2)
print("difference_set:", difference_set)
symmetric_difference_set = set1.symmetric_difference(set2)
print("symmetric_difference_set:", symmetric_difference_set)
```

```python
insersection_update_set = set1.intersection_update(set2)
print("insersection_update_set:", insersection_update_set)
```

```python
# set methods
set1 = {1, 2, 3, 4, 5}
set2 = {1, 2, 3, 4,4, 5, 6, 7, 8}
print(set1.issubset(set2))
print(set1.issuperset(set2))
```

```python
list = [1,2,2,3,4,4,5]
set_from_list = set(list)
print(set_from_list)
```

```python
# counting unique elements in the text
text = "in this text we have some duplicate words and we want to count the unique words"
words = text.split()
unique_words = set(words)
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```
