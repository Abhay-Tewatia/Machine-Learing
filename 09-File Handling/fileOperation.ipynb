File Operations- Read And Write Files

file handling is a crucial part of any programming language. Python provides built-in functions and methods to read from write to files, both text and binary. This lesson will cover the basics of file handling , including reading and writing text files and binary files.

```python
# Read a whole file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
# Read a file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip()) 
        # strip() is used to remove the newline character at the end of each line 
```

```python
## Write a file(Overwriting)

with open('example.txt', 'w') as file:
    file.write("Hello World!\n")
    file.write("This is a new line.")
```

```python
# Write a file(without Overwriting)
with open('example.txt', 'a') as file:
    file.write("Append operations taking place\n")
```

```python
# writing a list of lines to a file
lines = ['first line\n', 'Second line\n', 'Third line\n']
with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
# Binary Files

# Writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# Reading a binary file
with open('example.bin', 'rb') as file:
    content = file.read()
    print(content)
```

```python
# Read the content from source text file and write to a destination text file
# copying a text file

with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
# Read a text file and count the number of line, words, and characters in it
with open('example.txt', 'r') as file:
    content = file.read()

    lines = content.splitlines()
    words = content.split()
    characters = len(content)

    print(f"Number of lines: {len(lines)}")
    print(f"Number of words: {len(words)}")
    print(f"Number of characters: {characters}  ")
```
