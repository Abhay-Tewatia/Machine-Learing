“Decorators in Python are higher-order functions that wrap another function to extend its behavior without modifying its source code.”

🔁 4. Function inside Function

```python
def outer():
    def inner():
        print("Inner function")
    inner()

outer()
# 👉 Function ke andar function bana sakte ho
```

🔐 5. Closure (IMPORTANT)

```python
def outer():
    def inner():
        print("Hello from inner")
    return inner

fn = outer()
fn()
# 👉 Outer function return kar raha hai inner function
# 👉 Yeh closure hai
```

```python
def locker_banao(password):      # locker bana, password set kiya
    
    def check_karo(attempt):     # har baar check karo
        if attempt == password:
            print("✅ Locker khul gaya!")
        else:
            print("❌ Wrong password!")
    
    return check_karo            # locker de diya — password andar band hai

# Locker ban gaya — password "1234" andar lock ho gaya
mera_locker = locker_banao("1234")

# locker_banao() khatam ho gaya — but password yaad hai!
mera_locker("0000")   # ❌ Wrong password!
mera_locker("1234")   # ✅ Locker khul gaya!
```

```python
# decorator
def main_welcome(func):
    
    def sub_welcome_method():
        print("Welcome to the ai ml course")
        func()
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
def course_introduction():
    print("This is an advance python course")

course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("This is an advance python course")
```

```python
# # decorator
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called")
        func()
        print("something is happening after the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
say_hello()
```

```python
def total(*args):
    print(args)

total(1,2,3,4)
total(1,2,3,4,5)
```

```python
def info(**kwargs):
    print(kwargs)

info(name="Abhay", age=20)
info(name="Abhay Tewatia", age=23)
```

```python

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
              func(*args, **kwargs) 
        return wrapper 
    return decorator 
```

```python
@repeat(3)
def say_hello():
    print("Hello")
say_hello()
```

```python
@repeat(3)
def greet(name):
    print(f"Hello {name}")
greet("Abhay")
```

```python
@repeat(2)
def intro(name, age):
    print(f" I am {name},and i am  {age} year old")
intro("Abhay", 23)
```
