```python
# Har Iterator ke andar 2 methods hote hain:

# __iter__() → self return karta hai
# __next__() → agla element return karta hai
#              ya StopIteration raise karta hai
```

```python
# For loop ka actual kaam
_iterator = iter([1, 2, 3])     # Step 1: iterator banao

while True:                      # Step 2: loop
    try:
        n = next(_iterator)      # Step 3: next element lo
        print(n)                 # Step 4: kaam karo
    except StopIteration:        # Step 5: khatam?
        break                    #         band karo
```

```python
list = [1,2,34,5,6]
for i  in list:
    print(i)

print(type(list))
print(list)
```

```python
# Iterator
iterator = iter(list)
print(type(iterator))
```

```python
try:
    print(next(iterator))
except StopIteration:
    print("There are no element in the iterator")
```
