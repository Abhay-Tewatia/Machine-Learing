```python
### Assignment 1: Importing and Using Modules
# Import the `math` module and use it to calculate the square root of 25 and the sine of 90 degrees.

import math
sqrt_25 = math.sqrt(25)
sine_90 = math.sin(math.radians(90))  # Convert degrees to radians for the sine function

print("Square root of 25:", sqrt_25)
print("Sine of 90 degrees:", sine_90)
```

```python
## Lesson 5.2: Standard Library Overview
### Assignment 6: Working with the `os` Module

# Use the `os` module to create a new directory, list the contents of the current directory, and remove the newly created directory.

import os
# create a new directory
new_directory = "test_directory"
os.makedirs(new_directory, exist_ok=True)

# list the contents of the current directory
contents = os.listdir()
print("Contents of the current directory:")
for item in contents:
    print(item)

# remove the newly created directory
os.rmdir(new_directory)
```

```python

### Assignment 8: Working with the `math` Module

# Use the `math` module to calculate the greatest common divisor (GCD) of two numbers and the factorial of a number.

import math
# calculate the GCD of two numbers
num1 = 48
num2 = 18
gcd = math.gcd(num1, num2)
print(f"GCD of {num1} and {num2} is: {gcd}")

# calculate the factorial of a number
number = 5
factorial = math.factorial(number)
print(f"Factorial of {number} is: {factorial}")
```

```python
### Assignment 9: Working with the `datetime` Module

# Use the `datetime` module to print the current date, calculate the date 100 days from today, and determine the day of the week for a given date.

import datetime
# print the current date
current_date = datetime.date.today()
print("Current date:", current_date)

# calculate the date 100 days from today
date_100_days = current_date + datetime.timedelta(days=100)
print("Date 100 days from today:", date_100_days)

# determine the day of the week for a given date
day_of_week = date_100_days.strftime("%A")
print(f"The day of the week for {date_100_days} is: {day_of_week}")
```

```python
### Assignment 10: Working with the `random` Module
# Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.
import random
# generate a list of 5 random numbers between 1 and 50
random_numbers = [random.randint(1, 50) for _ in range(5)]
print("Random numbers:", random_numbers)

# shuffle the elements of the list
random.shuffle(random_numbers)
print("Shuffled numbers:", random_numbers)
```
