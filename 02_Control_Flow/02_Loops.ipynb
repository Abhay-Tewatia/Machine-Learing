```python
# while loop
# the while loop is used to execute a block of code repeatedly until a certain condition is met or becomes false. 
count = 0
while count <5:
    print(count)
    count += 1
```

```python
count = 0
while count%2 == 0:
    print(count)
    count += 1
    
```

```python
# looop control statements
# break statement
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
# continue statement
for i in range(10):
    if i%2==0:
        continue
    print(i)
```
